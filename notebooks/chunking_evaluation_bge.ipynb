# BGE semantic chunking evaluation

Compare 350/50, 250/40, and 150/30 chunking with `BAAI/bge-small-en-v1.5`. This uses the same 51 scored questions and source-level Recall@k/MRR definitions as the TF–IDF notebook, excluding Q46–Q52, Q58, and Q60.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
import pandas as pd

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'data').exists() and (PROJECT_ROOT.parent / 'data').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
from evaluate_bge import run_evaluation
pd.set_option('display.max_colwidth', 120)
PROJECT_ROOT
```

```python
result = run_evaluation(PROJECT_ROOT, batch_size=32)
result['run_info']
```

## Overall results

```python
summary = result['summary']
summary.style.format('{:.3f}').highlight_max(axis=0, color='#c6efce')
```

```python
ax = summary.plot(kind='bar', figsize=(10, 5), ylim=(0, 1), rot=0)
ax.set_title('BGE semantic retrieval metrics by chunking configuration')
ax.set_ylabel('Score')
ax.grid(axis='y', alpha=0.25)
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()
```
